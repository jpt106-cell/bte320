```python
import random #

decision = input("Do you want to play rock paper scissors? (yes or no) ")
while decision == "yes":
  computer_move = random.choice(['rock', 'paper','scissors'])
  player_move = input("Choose your pick (rock, paper, or scissors) ").lower()
  while computer_move == player_move:
    print(f"It's a tie! You both picked {player_move}. Play again.")
    computer_move = random.choice(['rock', 'paper', 'scissors'])
    player_move = input("Choose your new pick ")

  print(f"Computer played {computer_move}")
  print(f"You played {player_move}")
  if (computer_move == "rock" and player_move == "scissors"
      or computer_move == "scissors" and player_move == "paper"
      or computer_move == "paper" and player_move == "rock"):
    print("You lost!")
  else:
    print("You won!")
  decision = input("Do you want to play again? (yes or no) ")

```

```python
base_pay = 5

def parking_fee(hours):
    pay = base_pay + hours * 2.5

    if pay > 20:
        pay = 20.0
    if pay < 10:
        pay = 10.0

    return pay


print("min fee: $10  max fee: $20\n")

for i in range(1,9):
    fee = parking_fee(i)
    print(f"hours parked {i}    fee: ${fee}")
```
