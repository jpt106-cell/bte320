```python
#Rock Paper Scissors
import random

decision = input("Do you want to play rock paper scissors? (yes or no) ")

while decision == "yes":
  computer_move = random.choice(['rock', 'paper', 'scissors'])
  player_move = input("Choose your pick (rock, paper, or scissors) ").lower()

  print(f"Computer played {computer_move}")
  print(f"You played {player_move}")

  if computer_move == player_move:
    print("It's a tie!")
  elif (computer_move == "rock" and player_move == "scissors"
      or computer_move == "scissors" and player_move == "paper"
      or computer_move == "paper" and player_move == "rock"):
    print("You lost!")
  else:
    print("You won!")

  decision = input("Do you want to play again? (yes or no) ")
```

```python
#Recursive String Reversal
def reverse_string(s):
    if len(s) <= 1:
        return s
    return reverse_string(s[1:]) + s[0]

user_input = input("Enter a string to reverse: ")
print(reverse_string(user_input))
```

```python
#ROI
def roi(cost, revenue):
    if not isinstance(cost, (int, float)) or not isinstance(revenue, (int, float)):
        return "Error: cost and revenue must be numbers"

    if cost <= 0:
        return "Error: cost must be greater than 0"

    return (revenue - cost) / cost

cost = float(input("Enter cost: "))
revenue = float(input("Enter revenue: "))

print(roi(cost, revenue))
```

```python
#Loan
def loan_payment(principal, annual_rate, years):
    if principal <= 0 or annual_rate <= 0 or years <= 0:
        return "Error: all inputs must be positive"

    r = annual_rate / 12
    n = years * 12

    payment = (r * principal) / (1 - ((1 + r) ** -n))

    return payment

principal = float(input("Enter loan amount: "))
annual_rate = float(input("Enter annual interest rate as decimal: "))
years = int(input("Enter number of years: "))

print(f"Monthly payment is {loan_payment(principal, annual_rate, years):.2f}")
```

```python
#Quality Control Manager

def getValues(n):
    assert isinstance(n, int), "Input 'n' must be an integer!"
    assert n > 0

    items = []

    for i in range(n):
        try:
            items.append(float(input("Enter a value: ")))
        except:
            print("Invalid input")
            return getValues(n)

    return items

def calcAverage(alist):
    assert isinstance(alist, list), "Input parameter 'alist' must be a list"

    if len(alist) < 3:
        raise ValueError("the length of a list must be greater than 2.")

    return sum(alist[2:]) / len(alist[2:])

def main():
    try:
        n = int(input("Enter a number of items to be added to the list: "))
    except:
        print("Invalid input")
        return

    alist = getValues(n)
    avg = calcAverage(alist)
    print(f"avg: {avg} lbs.")

main()
```

```python
class Greeting:
    def __init__(self, first, last):
        self.first = first
        self.last = last

    def display(self, formal):
        if formal:
            print(f"{self.last}, {self.first}")
        else:
            print(self.first)

    def __str__(self):
        return f"{self.first} {self.last}"

first = input("Enter first name: ")
last = input("Enter last name: ")
formal = input("Display formally? (yes/no): ").lower() == "yes"

g = Greeting(first, last)
g.display(formal)
print(g)
```

```python
#Student
class Student:
    def __init__(self, name, number, courses):
        self.name = name
        self.number = number
        self.courses = courses

    def get_name(self):
        return self.name

    def get_courses(self):
        return self.courses

    def set_name(self, name):
        self.name = name

    def __len__(self):
        return len(self.courses)

    def enroll(self, course):
        if course not in self.courses:
            self.courses.append(course)

student1 = Student("Jake", 123, ["BTE 320", "CET 340"])

print(student1.get_name())
print(student1.get_courses())

student1.set_name("Jacob")
student1.enroll("MTH 210")
student1.enroll("BTE 320")

print(student1.get_name())
print(student1.get_courses())
print(len(student1))
```

```python
#Point and Rectangle Classes
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other_point):
        return ((self.x - other_point.x) ** 2 + (self.y - other_point.y) ** 2) ** 0.5


class Rectangle:
    def __init__(self, p1, p2, p3, p4):
        self.p1 = p1
        self.p2 = p2
        self.p3 = p3
        self.p4 = p4
        self.width = self.calc_width()
        self.height = self.calc_height()

    def calc_width(self):
        return self.p1.distance(self.p2)

    def calc_height(self):
        return self.p1.distance(self.p4)

    def area(self):
        return self.width * self.height


p1 = Point(0, 0)
p2 = Point(5, 0)
p3 = Point(5, 3)
p4 = Point(0, 3)

rect = Rectangle(p1, p2, p3, p4)

print("Width:", rect.calc_width())
print("Height:", rect.calc_height())
print("Area:", rect.area())
```

```python
class Employee:
    def __init__(self, name, wage, hours):
        self.name = name
        self.__wage = wage
        self.hours = hours

    def get_wage(self): return self.__wage
    def set_wage(self, w):
        assert w >= 0
        self.__wage = w

    def calculate_pay(self): return self.__wage * self.hours
    def display_info(self):
        print(f"{self.name} | Wage: ${self.__wage} | Hours: {self.hours} | Pay: ${self.calculate_pay()}")

class Manager(Employee):
    def __init__(self, name, wage, hours, bonus):
        super().__init__(name, wage, hours)
        self.bonus = bonus

    def calculate_pay(self): return super().calculate_pay() + self.bonus
    def display_info(self):
        super().display_info()
        print(f"Bonus: ${self.bonus}")

Employee("Benji Comer", 20, 35).display_info()
print()
Manager("Jake Tanapat", 30, 60, 500).display_info()
```
