```python
pct = .4
sales = float(input("Enter the amount of sales "))
cogs = pct*sales
gross_profit = sales-cogs
print(f"Total amount of COGS is {cogs} \nWhich means gross profit is {gross_profit}")
taxrate= float(input("What is the tax rate? (%) "))
taxes = gross_profit*(taxrate/100)
net_income = gross_profit-taxes
print(f"Taxes equal {taxes} and netincome is {net_income}\n")

print(f"Sales: {sales} \nCost of Goods sold: {cogs} \nGross Profit: {gross_profit}\nTaxes: {taxes}\nNet income: {net_income}")
```

```python
num1 = float(input("Enter the first number "))
num2 = float(input("Enter the second number "))
operator = (input("Enter the operator you want to use "))
if operator == "+":
  result = num1 + num2
elif operator == "-":
  result = num1 - num2
elif operator == "*":
  result = num1*num2
elif operator == "/":
  result = num1/num2
else:
  print("Invalid operator")

print(f"Your answer is {result}")
```
